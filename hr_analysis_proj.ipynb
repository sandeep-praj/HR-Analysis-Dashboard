```python
miss = df.isnull().sum().sort_values(ascending=False)
miss
```

```python
fill_missing_value = ['YearsWithCurrManager','EnvironmentSatisfaction','JobSatisfaction']

df[fill_missing_value] = df[fill_missing_value].fillna(
    df[fill_missing_value].mean())
```

```python
# delete a column over18 
df.drop(columns=['Over18'],inplace=True)

```

```python
fil_mis_va = df['Gender'] = df['Gender'].fillna(df['Gender'].mode().loc[0])
fil_mis_va
```

```python
re_nul = df.dropna(inplace=True)
re_nul
```

```python
categorical_columns = [
    "AgeGroup",
    "Attrition",
    "BusinessTravel",
    "Department",
    "EducationField",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "SalarySlab"
]

for col in categorical_columns:
    df[col] = df[col].astype("category")
```
